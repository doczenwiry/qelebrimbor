```python
import os

os.environ["EXTRA_CLING_ARGS"] = "-Wno-nonportable-include-path"
```

```python
from ogdf_python import cppinclude, ogdf

cppinclude("ogdf/basic/graph_generators/randomized.h")
cppinclude("ogdf/layered/SugiyamaLayout.h")
```

```python
# Taken from usage section of https://github.com/ogdf/ogdf-python

G = ogdf.Graph()
ogdf.setSeed(1)
ogdf.randomPlanarTriconnectedGraph(G, 20, 40)
GA = ogdf.GraphAttributes(G, ogdf.GraphAttributes.all)

for n in G.nodes:
    GA.label[n] = "N%s" % n.index()
    GA.height[n] = 80
    GA.width[n] = 80

SL = ogdf.SugiyamaLayout()
SL.call(GA)
GA
```
