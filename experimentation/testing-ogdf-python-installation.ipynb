```python
from ogdf_python import cppinclude, ogdf

cppinclude("ogdf/basic/graph_generators/randomized.h")
cppinclude("ogdf/layered/SugiyamaLayout.h")
cppinclude("ogdf/energybased/FMMMLayout.h")
cppinclude("ogdf/energybased/fmmm/FMMMOptions.h")
cppinclude("ogdf/planarity/PlanarizationLayout.h")
cppinclude("ogdf/orthogonal/OrthoLayout.h")
```

```python
# Taken from usage section of https://github.com/ogdf/ogdf-python
G = ogdf.Graph()
ogdf.setSeed(1)
ogdf.randomPlanarTriconnectedGraph(G, 20, 40)

attributes = ogdf.GraphAttributes(G, ogdf.GraphAttributes.all)

for n in G.nodes:
    attributes.label[n] = "N%s" % n.index()
    attributes.shape[n] = ogdf.Shape.Ellipse
    attributes.height[n] = 20
    attributes.width[n] = 20

algorithm = ogdf.SugiyamaLayout()
algorithm.call(attributes)
attributes
```

```python
attributes = ogdf.GraphAttributes(G, ogdf.GraphAttributes.all)

for n in G.nodes:
    attributes.label[n] = "N%s" % n.index()
    # attributes.shape[n] = ogdf.Shape.Ellipse

algorithm = ogdf.FMMMLayout()
algorithm.useHighLevelOptions(True)
algorithm.unitEdgeLength(40.0)
algorithm.newInitialPlacement(True)
algorithm.qualityVersusSpeed(ogdf.FMMMOptions.QualityVsSpeed.GorgeousAndEfficient)

algorithm.call(attributes)
attributes
```

```python
attributes = ogdf.GraphAttributes(G, ogdf.GraphAttributes.all)

for n in G.nodes:
    attributes.label[n] = "N%s" % n.index()
    attributes.shape[n] = ogdf.Shape.Ellipse
    attributes.height[n] = 80
    attributes.width[n] = 80


algorithm = ogdf.PlanarizationLayout()
layout = ogdf.OrthoLayout()
layout.separation(150)
layout.cOverhang(1.0)
layout.__python_owns__ = False  # Needed to avoid C++ segmentation fault when object is destroyed
algorithm.setPlanarLayouter(layout)

algorithm.call(attributes)

attributes
```
